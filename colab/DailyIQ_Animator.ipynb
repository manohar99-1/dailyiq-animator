# DailyIQ Animator
Type a topic, run all cells, get an animated Short.

1. Run **Setup** (about 2 minutes, once).
2. Add a free API key (Groq, OpenRouter or Gemini) in the **Keys** cell.
3. Set your topic and run **Generate**.

```
#@title Setup
!git clone -q https://github.com/manohar99-1/dailyiq-animator.git 2>/dev/null || true
%cd dailyiq-animator
!apt-get -qq install -y ffmpeg > /dev/null
!pip -q install -r requirements.txt
!python -m playwright install --with-deps chromium > /dev/null 2>&1
print('ready')
```

```
#@title Keys (any one is enough)
import os
os.environ['GROQ_API_KEY'] = ''        #@param {type:"string"}
os.environ['OPENROUTER_API_KEY'] = ''  #@param {type:"string"}
os.environ['GEMINI_API_KEY'] = ''      #@param {type:"string"}
```

```
#@title Generate
topic = "What if UPI stopped working?"  #@param {type:"string"}
voice = ""  #@param {type:"string"}
import shlex
v = f"--voice {shlex.quote(voice)}" if voice else ""
!python -m animator.cli generate --topic "{topic}" --out output {v}
```
